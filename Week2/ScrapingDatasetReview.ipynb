## **Import Library**

Kita memanggil library yang sudah diinstal dan menentukan target aplikasi serta jumlah data.

```python
from google_play_scraper import reviews, Sort
import pandas as pd

app_id = 'com.discord'
target_count = 20000
```

## **3. Execution Scraping (Looping & Batching)**

Agar pengambilan data 20.000 baris berjalan lancar, kita menggunakan sistem batch (pengambilan bertahap 200 ulasan sekali tarik) dan continuation token agar pengambilan data menyambung dari ulasan terakhir.

```python
all_reviews = []
continuation_token = None

print(f"Memulai scraping ulasan Indonesia untuk {app_id}...")

while len(all_reviews) < target_count:
    # Mengambil ulasan dengan filter Bahasa Indonesia (id) dan wilayah Indonesia (id)
    result, continuation_token = reviews(
        app_id,
        lang='id',          # Bahasa Indonesia
        country='id',       # Wilayah Indonesia
        sort=Sort.NEWEST,   # Ulasan terbaru
        count=200,          # Batch size per penarikan
        continuation_token=continuation_token
    )

    all_reviews.extend(result)
    print(f"Berhasil mengambil {len(all_reviews)} ulasan...")

    # Berhenti jika ulasan sudah habis di server sebelum mencapai target
    if not continuation_token:
        break

# Memastikan data tepat 20.000 ulasan
df = pd.DataFrame(all_reviews).head(target_count)
```

## **4. Data Selection & Saving**

Kita memilih kolom yang penting saja (seperti content dan score) lalu menyimpannya ke dalam file CSV.

```python
# Menyimpan kolom utama: review (content), rating (score), dan waktu (at)
df_final = df[['userName', 'score', 'at', 'content']]

# Simpan ke CSV
df_final.to_csv("ReviewAppDiscord.csv", index=False)
print(f"Data Berhasil Disimpan! Total: {len(df_final)} baris.")
```
