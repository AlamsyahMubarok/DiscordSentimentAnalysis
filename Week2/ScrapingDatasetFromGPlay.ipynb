## **Import Library**

```python
from google_play_scraper import reviews, Sort
import pandas as pd
```

## **Scraping Review Discord**

```python
all_reviews = []
batch_size = 200
limit = 15000  # Tentukan batas maksimal data yang diinginkan

# Scraping awal
result, continuation_token = reviews(
    'com.discord',
    lang='en',
    country='us',
    sort=Sort.NEWEST,
    count=batch_size
)

all_reviews.extend(result)

# Loop dengan tambahan pengecekan jumlah data
while continuation_token and len(all_reviews) < limit:

    result, continuation_token = reviews(
        'com.discord',
        continuation_token=continuation_token
    )

    if not result:
        break

    all_reviews.extend(result)

    # Print status agar kamu tahu progresnya
    print(f"Total review scraped: {len(all_reviews)}")

# Memastikan data tidak lebih dari limit jika batch terakhir kelebihan
all_reviews = all_reviews[:limit]
print(f"Selesai! Total data akhir: {len(all_reviews)}")
```

## **Convert To Dataframe**

```python
df = pd.DataFrame(all_reviews)
```

## **Data Cleaning**

```python
df = df[['content', 'score', 'thumbsUpCount', 'reviewCreatedVersion', 'at', 'appVersion']]
```

## **Export to CSV**

```python
df.to_csv("Review_App_Discord.csv", index=False)
```
